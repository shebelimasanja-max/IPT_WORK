```python
import pandas as pd
```

```python
data={
    "Assessment":["Good","Average","Good","Poor","Good",
                  "Average","Good","Poor","Average","Good"],
    "Assignment":["Yes","Yes","No","No","Yes","No","No","Yes","No","Yes"],
    "Project":["Yes","No","Yes","No","Yes","Yes","No","Yes","No","Yes"],
    "Result (%)":[95,70,75,45,98,80,75,65,58,89]
}
data=pd.DataFrame(data)
```

```python
data
```

```python
# Standard deviation
std_gen=round(data["Result (%)"].std(),4)
std_gen
```

```python
std_reduction_list=[]
```

```python
def std_func(list_,col):
    total_attribute_std=[]
    for val in list_:
        a=data[col]==val
        print(val)
        temp_table=data.loc[a]
        print(temp_table)
        std_=round(temp_table["Result (%)"].std(),4)
        print(f"std: {std_}")
        n=temp_table[col].count()
        std_val=(n/10)*std_
        total_attribute_std.append(std_val)
        # print(f"std: {total_attribute_std}")
        print()
    sum_std=round(sum(total_attribute_std),4)
    print(f"weighted standard deviation: {sum_std}")
    std_reduction=round((std_gen-sum_std),4)
    print(f"Standard deviation reduction {std_reduction}")
    std_reduction_list.append({
        "Attribute":col,
        "std_reduction":std_reduction
    })
    display(pd.DataFrame(std_reduction_list))
    print()
```

```python
# For Assessment column
std_func(["Good","Average","Poor"],"Assessment")
```

```python
# For Assignemt column
std_func(["Yes","No"],"Assignment")
```

```python
std_func(["Yes","No"],"Project")
```
